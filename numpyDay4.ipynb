Practice programs based on numpy

```python
import numpy as np
```

```python
#1.Create a 3x3 matrix with values from 1 to 9.
mat=np.arange(1,10).reshape(3,3)
mat
```

```python
#2.Replace all odd numbers in an array with -1.
arr=np.arange(10)
arr1=np.where(arr%2==0,arr,-1)
arr1
```

```python
#3.Create an array of shape (4, 4) filled with a single repeating number.
arr=np.ones([4,4])*5
arr
```

```python
#4.Create a diagonal matrix using a list.
l=[1,2,3,4,5]
mat1=np.diag(l)
mat1
```

```python
#5.Generate an identity matrix using NumPy.
id=np.eye(5)
id
```

```python
#6.Create a boolean mask for values > 50 in a random array of 20 ints (1–100).
arr=np.random.randint(1,100,20)
print(arr)
arr=arr>50
print(arr)
```

```python
#7.Reshape a 1D array of 16 values into a 4x4 matrix.
arr=np.arange(16)
arr.reshape(4,4)
```

```python
#8.Perform element-wise multiplication between two 1D arrays.
arr1=np.arange(5)
arr2=np.arange(5,10)
mul=arr1*arr2
print(arr1,arr2,'\n',mul)
```

```python
#9.Swap two rows in a 2D array.
arr=np.array([[1,2,3],[4,5,6]])
a=arr[[0,1]]=arr[[1,0]]
a
```

```python
#10.Replace NaN values in an array with the column mean.
arr=np.array([[1,2,3,np.nan],[5,6,np.nan,8],[9,np.nan,11,12]])
print(arr)
col_mean=np.nanmean(arr,axis=0)
print(f"col_mean{col_mean}")
nan_index=np.where(np.isnan(arr))
arr[nan_index]=np.take(col_mean,nan_index[1])
arr
```

```python
#11.Compute the dot product of two matrices
mat1=np.array([[1,2],[3,4]])
mat2=np.array([[5,6],[7,8]])
print(f'mat1:\n{mat1}\nmat2:\n{mat2}')
dot_product=np.dot(mat1,mat2)
print("dot product:\n",dot_product)
```

```python
#12.Find unique elements and their counts in an array
arr = np.array([1, 2, 2, 3, 4, 4, 4])
unique=np.unique(arr)
count=np.unique(arr,return_counts=True)
print("unique values:",unique)
print("count:",count)
```

```python
#13.Stack two arrays vertically and horizontally
arr1=np.array([[1,2],[3,4]])
arr2=np.array([[5,6],[7,8]])
print("vertically stack:\n",np.vstack((arr1,arr2)))
print("horizontally stack:\n",np.hstack((arr1,arr2)))
```

```python
#14.Find standard deviation of each row in a 2D array
arr=np.arange(9).reshape(3,3)
std=np.std(arr,axis=1)
std
```

```python
#15.Normalize a NumPy array (0 to 1 scale).
arr=np.array([10,20,30,40,50])
print("original:",arr)
normalised=(arr-arr.min())/(arr.max()-arr.min())
print("after normalised:",normalised)
```
