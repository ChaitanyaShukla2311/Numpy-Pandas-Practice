```python
import numpy as np
```

```python
#1.create 5X5 matrix where each row is [1,2,3,4,5] using broadcasting
row=[1,2,3,4,5]
mat=np.ones((5,1))*row
mat
```

```python
#2.replace all negative numbers with their positive in random array 
arr=np.random.randint(-100,100,10)
print("oringinal array:",arr)
positive=np.where(arr<0,-arr,arr)
print("positive:",positive)
```

```python
#3.find top 3 largest elements and their indices
arr=np.random.randint(0,100,10)
print("array:",arr)
ind=np.argsort(arr)[-3:][::-1]
print("top 3 largest value's index:",ind)
val=arr[ind]
print("top 3 largest value",val)
```

```python
#4.print 4X4 matrix and find maximum of digonal element and its index
mat=np.random.randint(1,50,(4,4))
print("matrix:\n",mat)
diagonal=np.diag(mat)
print(f"maximum of digonal element is:{np.max(diagonal)} and its index is:{np.argmax(diagonal)}")
```

```python
#5.set all elements below column mean to zero
mat=np.random.randint(1,100,(4,5))
col_means=mat.mean(axis=0)
result=np.where(mat<col_means,0,mat)
print("original:\n",mat)
print("column means:",col_means)
print("below column mean to zero\n",result)
```

```python
#6.Extract  every 3rd element in reverse
arr=np.random.randint(1,100,20)
print("array:\n",arr)
print("every 3rd element in reverse:\n",arr[::-3])
```

```python
#7.find distance of each element from the mean
arr=np.random.randint(1,100,10)
m=np.mean(arr)
d=arr-m
print("array:\n",arr)
print("mean:\n",m)
print("distance from mean:\n",d)
```
