Practice programs pandas  

```python
import pandas as pd
import numpy as np
```

```python
#1.Create DataFrame from a dictionary
df=pd.DataFrame({'name':['ram','diya','om'],
                 'age':[20,19,21],
                 'gender':['male','female','male']})
df
```

```python
#2.Rename columns of a DataFrame.
df1=df.rename(columns={'name':'fullname'})
df1
```

```python
#3. Sort DataFrame by multiple columns
df.sort_values(by=['gender','age'])
```

```python
#4.Fill missing values with forward fill method.
df2=pd.DataFrame({'A': [1, np.nan, 3], 'B': [np.nan, 5, 6]})
dff=df2.fillna(method='ffill')
dff
```

```python
#5.Create a new column that is sum of two existing columns.
data=pd.DataFrame({'Math': [80, 90], 'Science': [85, 95]})
data['total']=data['Math']+data['Science']
data
```

```python
#6.Drop duplicate rows based on one column
data=pd.DataFrame({'Name': ['Alice', 'Bob', 'Alice'], 'Age': [25, 30, 25]})
print(data)
d=data.drop_duplicates(subset='Name')
d
```

```python
#7.Filter rows where 'Name' contains the letter 'a'
data[data['Name'].str.contains('a',case=False)]
```

```python
#8. Group by a category and calculate the mean
df=pd.DataFrame({'Department': ['HR', 'IT', 'HR'], 'Salary': [50000, 60000, 55000]})
grp=df.groupby(df['Department']).mean()
grp
```

```python
#9.Convert column datatype to integer
dff=pd.DataFrame({'Age': ['25', '30', '22']})
dff['Age'] = dff['Age'].astype(int)
dff
```

```python
#10.Replace all 'active' values with 'enabled'
d=pd.DataFrame({'Status': ['active', 'inactive', 'active']})
d['Status'] = d['Status'].replace('active', 'enabled')
d
```

```python
#11.Convert date column and extract year/month
df=pd.DataFrame({'JoinDate': ['2021-01-10', '2022-03-15']})
df['JoinDate'] = pd.to_datetime(df['JoinDate'])
df['Year'] = df['JoinDate'].dt.year
df['Month'] = df['JoinDate'].dt.month
df
```
