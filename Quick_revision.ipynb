| `count()`   | `list.count(item)`         | Counts how many times item occurs                | `my_list.count(2)`       |

# class

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return f"Hello, {self.name}"

p1 = Person("Chaitanya")
print(p1.greet())  # Output: Hello, Chaitanya
```

```python
#**Inheritance**
class Parent:
    def show(self):
        print("Parent method")

class Child(Parent):
    def greet(self):
        print("Child method")

obj = Child()
obj.show()   # Inherited
obj.greet()  # Own method
```

# t test,z-test,ANOVA test
#this is for numerical features

```python
import numpy as np
a=np.linspace(90,150,40)
a_list=list(a)
print(a_list)
```

```python
b=np.linspace(80,190,40)
b_list=list(b)
print(b_list)
```

```python
!pip install statsmodels
```

```python

from statsmodels.stats.weightstats import ztest
stat,p=ztest(a_list,b_list)  #output: z-stats and p-value
print(f'Statistics={stat.round(3)} ,p={p.round(3)}')

#interpret
alpha=0.05
if p> alpha:
    print('Same distribution (fail to reject H0)')
else:
    print('Different distribution (reject H0)')
```

```python
#ttest for two independent samples

from scipy.stats import ttest_ind
data1=5 * np.random.randn(100) +50
data2=5 * np.random.randn(100) +51

stat,p=ttest_ind(data1,data2)  #output: t-stats and p-value
print(f'Statistics={stat.round(3)} ,p={p.round(3)}')

#interpret
alpha=0.05
if p> alpha:
    print('Same distribution (fail to reject H0)')
else:
    print('Different distribution (reject H0)')
```

```python
#ttest for paired samples

from scipy.stats import ttest_rel

data1=5 * np.random.randn(100) +50
data2=5 * np.random.randn(100) +51

stat,p=ttest_rel(data1,data2)  #output: t-stats and p-value
print(f'Statistics={stat.round(3)} ,p={p.round(3)}')

#interpret
alpha=0.05
if p> alpha:
    print('Same distribution (fail to reject H0)')
else:
    print('Different distribution (reject H0)')
```

```python
#ANOVA test is for more than two samples

from scipy.stats import f_oneway

data1=5 * np.random.randn(100) +50
data2=5 * np.random.randn(100) +51
data3=5 * np.random.randn(100) +52


stat,p=f_oneway(data1,data2,data3)  #output: t-stats and p-value
print(f'Statistics={stat.round(3)} ,p={p.round(3)}')

#interpret
alpha=0.05
if p> alpha:
    print('Same distribution (fail to reject H0)')
else:
    print('Different distribution (reject H0)')
```
