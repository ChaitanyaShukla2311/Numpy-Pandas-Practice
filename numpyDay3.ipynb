# programs based on numpy basics

```python
import numpy as np
```

```python
#1.create a 1D array using linspace and reshape it into a 3X5 matrix,replace all values <5 with 0
arr=np.linspace(1,15,15).reshape(3,5)
print(arr)
replace=np.where(arr<5,0,arr)
print("\n",replace)
```

```python
#2.create a 5X5 identity matrix and multiply it with an array[1,2,3,4,5]
arr=np.array([1,2,3,4,5])
mat=np.eye(5)
print("identity matrix:\n",mat)
print("After multiplying by array:\n",mat*arr)
```

```python
#3.product prices: filter all prices between 500 and 1000
prices=np.random.randint(100,2000,20)
print("all prices:\n",prices)
f=prices[(prices>=500) & (prices<=1000)]
print("after applying filter for prices betn 500 to 1000:\n",f)
```

```python
#4.count how many votes >8 in each row
votes=np.random.randint(1,11,(5,4))  #5 people 4 questions
print("votes:\n",votes)
count=np.sum(votes>8,axis=1)
print("count:\n",count)
```

```python
#5.find product with highest total rating
rate=np.random.randint(1,11,(8,5)) #8 users 5 products
print("ratings :\n",rate)
print("Total:",rate.sum(axis=0))
high=np.argmax(rate.sum(axis=0))
print("index of highest rated product:",high)
```

```python
#6.apply 10% discount where price > 1000
prices=np.random.randint(500,2000,10)
discounted=np.where(prices>1000,prices*0.9,prices)
print("oringinal prices:\n",prices)
print("discounted prices:\n",discounted)
```

```python
#7.count students who are present all 5 days 
attendence=np.random.randint(0,2,(10,5))  #0-absent 1-present 
whole=np.all(attendence==1,axis=1)
print("Students attendence:\n",attendence)
print("present all 5 days:\n",np.where(whole==True))
```

```python
#8.find the outliers(>mean+2*std) and replace with 0
data=np.random.normal(50,10,(6,6))
print("Data:\n",data)
mean=np.mean(data)
std=np.std(data)
filtered=np.where(data>mean+2*std,0,data)
print("outliers are denoted by zero:\n",filtered)
```

```python
#9.print pass or fail for 10 students based on 5 subjects marks
marks=np.random.randint(20,100,(10,5))
result=np.where(np.any(marks<30,axis=1),'Fail','Pass')
print("marks:\n",marks)
print("Result:\n",result)
```

```python
#10.mark all expenses >25000 as 'Over budget'
exp=np.random.randint(1000,50000,(5,6)).astype(object)
exp[exp.astype(int)>25000]="OverBudget"
print("expenses matrix\n",exp)
```
