# 4
Calculate Average Marks for Each Subject

```python
print("Average marks for Math:",data['Math'].mean())
print("Average marks for Science:",data['Science'].mean())
print("Average marks for English:",data['English'].mean())
```

# 5
Add a New Column Total and Percentage
Total = Math + Science + English
Percentage = Total / 3
Add them as new columns to the DataFrame

```python
data['Total']=data['Math']+data['Science']+data['English']
data['Percentage']=(data['Total']/3).round(2)
data
```
